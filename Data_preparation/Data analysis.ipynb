```python
# Import de librairie
import pandas as pd
import numpy as np
```

# Database dechets par habitants

```python
annees = [str(annee) for annee in range(2003, 2022)]
```

```python
Data_dechets= []
for annee in annees:
    data_d = pd.read_excel(
        "Dechet_par_habitant.xls",
        sheet_name=annee,        
        skiprows=1,              
        usecols="A:B",          
        header=None,             
        names=["Commune", "Dechet_par_habitant_kg"]
    )
    data_d["Année"] = int(annee)     
    Data_dechets.append(data_d) 

Data_dechets_final = pd.concat(Data_dechets, ignore_index=True)

print(Data_dechets_final)
print(type(Data_dechets_final))

```

```python
Data_dechets_final = Data_dechets_final [["Commune", "Année", "Dechet_par_habitant_kg"]]
Data_dechets_final = Data_dechets_final.sort_values(by=["Commune", "Année"])

Data_dechets_final = Data_dechets_final.reset_index(drop=True)

print(Data_dechets_final)

```

```python

dechet_moyennes_par_annee = Data_dechets_final.groupby('Année')['Dechet_par_habitant_kg'].mean().reset_index()
#print(dechet_moyennes_par_annee )

# Créer le dataframe des moyennes
dechet_moyennes =dechet_moyennes_par_annee.copy()
dechet_moyennes['Commune'] = 'Moyenne' 

Data_dechets_final = pd.concat([Data_dechets_final, dechet_moyennes], ignore_index=True)
print(Data_dechets_final)
Data_dechets_final = Data_dechets_final.sort_values(['Commune','Année'])
print(Data_dechets_final)

nombre_communes = Data_dechets_final["Commune"].nunique()
print(f"Nombre de communes : {nombre_communes}")

Data_dechets_final.to_csv("donnees_dechet_par_habitant.csv")
```

# Database densité de la population

```python
Data_densite = pd.read_excel("densité_de_la_population.xls")

Data_densite_reviewed = Data_densite.melt(id_vars=[Data_densite.columns[0]],var_name="Année", value_name="Densité_km2")
Data_densite_reviewed
Data_densite_reviewed.rename(columns={Data_densite.columns[0]: "Commune"}, inplace=True)

Data_densite_reviewed = Data_densite_reviewed.sort_values(by=["Commune", "Année"])

Data_densite_reviewed = Data_densite_reviewed .reset_index(drop=True)
Data_densite_reviewed

```

```python
densite_moyennes_par_annee = Data_densite_reviewed.groupby('Année')['Densité_km2'].mean().reset_index()

# Créer le dataframe des moyennes
densite_moyennes =densite_moyennes_par_annee.copy()
densite_moyennes

densite_moyennes['Commune'] = 'Moyenne' 
densite_moyennes
Data_densite_final = pd.concat([Data_densite_reviewed, densite_moyennes], ignore_index=True)
Data_densite_final = Data_densite_final.sort_values([ 'Commune','Année'])
print(Data_densite_final)
Data_densite_final.to_csv("donnee_densite_population.csv")
```

# Database logement subventionné 

```python
annees = [str(annee) for annee in range(2009, 2025)]

Data_logement= []
for annee in annees:
    data_l = pd.read_excel(
        "Logement subventionnée.xls",
        sheet_name=annee,        
        skiprows=1,              
        usecols="A:B",          
        header=None,             
        names=["Commune", "Nombre_de_logements"]
    )
    data_l["Année"] = int(annee)     
    Data_logement.append(data_l) 

Data_logement_final = pd.concat(Data_logement, ignore_index=True)

Data_logement_final = Data_logement_final [["Commune", "Année", "Nombre_de_logements"]]



Data_logement_final = Data_logement_final.sort_values(by=["Commune", "Année"])


Data_logement_final = Data_logement_final.reset_index(drop=True)

print(Data_logement_final)
```

```python
print(Data_logement_final['Nombre_de_logements'].dtype)
Data_logement_final['Nombre_de_logements'] = pd.to_numeric(Data_logement_final['Nombre_de_logements'], errors='coerce')
print(Data_logement_final['Nombre_de_logements'].dtype)


logement_moyennes_par_annee = Data_logement_final.groupby('Année')['Nombre_de_logements'].mean().reset_index()

# Créer le dataframe des moyennes
logement_moyennes =logement_moyennes_par_annee.copy()

logement_moyennes['Commune'] = 'Moyenne' 
print(logement_moyennes)
Data_logement_final = pd.concat([Data_logement_final, logement_moyennes], ignore_index=True)
Data_logement_final = Data_logement_final.sort_values(['Commune','Année'])
print(Data_logement_final)
Data_logement_final.to_csv("donnee_logement_subventionnee.csv")
```

# Revenu brute 

```python
annees = [str(annee) for annee in range(2010, 2019)]

Data_revenu= []
for annee in annees:
    data_r = pd.read_excel(
        "revenu_brut_annuel.xls",
        sheet_name=annee,        
        skiprows=1,              
        usecols="A:C",          
        header=None,             
        names=["Commune", "Médiane_couples_mariés", "Médiane_celibataire"]
    )
    data_r["Année"] = int(annee)     
    Data_revenu.append(data_r) 

Data_revenu_final = pd.concat(Data_revenu, ignore_index=True)

Data_revenu_final = Data_revenu_final [["Commune", "Année", "Médiane_couples_mariés", "Médiane_celibataire"]]



Data_revenu_final = Data_revenu_final.sort_values(by=["Commune", "Année"])


Data_revenu_final = Data_revenu_final.reset_index(drop=True)
Data_revenu_final
```

```python
indicateurs = ['Médiane_couples_mariés', 'Médiane_celibataire']
revenu_moyennes_par_annee = Data_revenu_final.groupby('Année')[indicateurs].mean().reset_index()
revenu_moyennes_par_annee

# Créer le dataframe des moyennes
revenu_moyennes =revenu_moyennes_par_annee.copy()

revenu_moyennes['Commune'] = 'Moyenne' 
print(len(revenu_moyennes))

Data_revenu_final = pd.concat([Data_revenu_final, revenu_moyennes], ignore_index=True)
Data_revenu_final =Data_revenu_final.sort_values([ 'Commune','Année'])
print(Data_revenu_final)
Data_revenu_final.to_csv("donnee_revenu_brut_annuel.csv")
```

# Données environnementales 

```python
Data_env =pd.read_csv("energyreporter_municipality_historized.csv")
Data_env_GE = Data_env.loc[Data_env["canton"]=="GE"]
Data_env_GE = Data_env_GE[["municipality","energyreporter_date",
                           "electric_cars_per_charging_spot","elec_consumption_mwh_per_year_per_capita",
                           "renelec_production_mwh_per_year_per_capita"]]
Data_env_GE
```

```python
Data_env_final= Data_env_GE.dropna().reset_index(drop=True)
Data_env_final = Data_env_final.rename(columns={'municipality': 'Commune'})
Data_env_final = Data_env_final.rename(columns={'energyreporter_date': 'Année'})
Data_env_final.sort_values(by=["Commune", "Année"])
print(len(Data_env_final))
```

```python
indicateurs_env = ['electric_cars_per_charging_spot', 'elec_consumption_mwh_per_year_per_capita', "renelec_production_mwh_per_year_per_capita"]
indicateurs_env_moyennes_par_annee = Data_env_final.groupby('Année')[indicateurs_env].mean().reset_index()
print(indicateurs_env_moyennes_par_annee)

# Créer le dataframe des moyennes
indicateurs_env_moyennes =indicateurs_env_moyennes_par_annee.copy()

indicateurs_env_moyennes['Commune'] = 'Moyenne' 
print(len(indicateurs_env_moyennes))

Data_env_final = pd.concat([Data_env_final, indicateurs_env_moyennes], ignore_index=True)
Data_env_final =Data_env_final.sort_values([ 'Commune','Année'])


Data_env_final.rename(columns={
    "electric_cars_per_charging_spot": "Voitures_electriques_par_borne",
    "elec_consumption_mwh_per_year_per_capita": "Consommation_electrique_annuelle_MWh_par_habitant",
    "renelec_production_mwh_per_year_per_capita": "Production_electrique_renouvelable_annuelle_MWh_par_habitant"
})

nombre_communes =Data_env_final["Commune"].nunique()
print(f"Nombre de communes : {nombre_communes}")

Data_env_final.to_csv("donnee_environement.csv")
Data_env_final
```
